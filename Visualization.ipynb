```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.lines as mlines
import numpy as np
from matplotlib.backends.backend_pdf import PdfPages

# ── Flagged targets ────────────────────────────────────────────────────────────
# Targets listed here are written to their OWN PDF (flagged_output_pdf) and are
# EXCLUDED from the main PDF.
#FLAGGED_TARGETS = ['glgC_5_UTR','csrB_sRNA','dcgZ_5_UTR','relA_5_UTR','ycdT_5_UTR','miaA_hfq_UTR','cstA_5_UTR','pgaA_5_UTR','nhaA_5_UTR','nhaA_nhaR_UTR',
                  #'flhD_5_UTR','iraD_5_UTR','flhD_5_UTR','moaA_5_UTR','sdiA_5_UTR','rpoE_5_UTR']

#FLAGGED_TARGETS = ['mcaS_sRNA', 'sgrS_sRNA','micL_sRNA','sibA_sRNA','chiX_sRNA','rprA_sRNA','rprA_sRNA','dsrA_sRNA','rnpB_sRNA','glmZ_sRNA','sibD_sRNA','fnrS_sRNA','cyaR_sRNA','gcvB_sRNA','sibC_sRNA']
FLAGGED_TARGETS = ['ppk_5_UTR','cmk_rpsA_UTR','skp_5_UTR','uxuA_5_UTR','mdtE_5_UTR','ompR_5_UTR','ribB_5_UTR','patA_5_UTR','metC_5_UTR','proS_5_UTR']
#FLAGGED_TARGETS = ['fecA_5_UTR','gmk_5_UTR','phoB_5_UTR']
#FLAGGED_TARGETS = ["acs_5_UTR","adeP_5_UTR","adhP_maeA_UTR","aroA_ycaL_UTR","bfr_bfd_UTR","cbpA_5_UTR","cbpA_5_UTR","clpS_clpA_UTR","cmk_rpsA_UTR","crp_5_UTR","csiD_5_UTR","csrA_5_UTR","csrA_5_UTR","cstA_5_UTR","cyoA_5_UTR","cysJ_5_UTR","ddlA_5_UTR","dgcZ_5_UTR","dgcZ_5_UTR","dgcZ_5_UTR","dnaK_5_UTR","dppA_5_UTR","dps_5_UTR","dsrB_5_UTR","elaB_elaA_UTR","elaB_elaA_UTR","eptC_ppc_UTR","eptC_ppc_UTR","fbp_5_UTR","fdoH_fdoG_UTR","fepA_5_UTR","fimA_5_UTR","fimA_5_UTR","flu_yeeR_UTR","frdA_5_UTR","frdA_5_UTR","frdB_5_UTR","fucO_fucA_UTR","fumA_5_UTR","fumA_5_UTR","gadB_5_UTR","glgB_5_UTR","glgB_5_UTR","glgB_5_UTR","glgC_5_UTR","glgC_5_UTR","glgC_5_UTR","glmS_glmU_UTR","glmS_glmU_UTR","glnH_5_UTR","glnS_5_UTR","gpt_5_UTR","groS_5_UTR","groS_5_UTR","groS_groL_UTR","groS_groL_UTR","groS_groL_UTR","gstA_5_UTR","hcaT_hcaR_UTR","hemX_hemD_UTR","hemX_hemD_UTR","hflX_hflK_UTR","hflX_hflK_UTR","hyfJ_hyfR_UTR","icd_5_UTR","icd_5_UTR","iraD_5_UTR","iscR_5_UTR","iscS_iscR_UTR","ldtA_cobT_UTR","lrp_5_UTR","lrp_ftsK_UTR","lsrB_lsrF_UTR","lsrB_lsrF_UTR","lsrG_tam_UTR","ltaE_5_UTR","maeA_sra_UTR","maeA_sra_UTR","maeA_sra_UTR","maeB_5_UTR","manX_5_UTR","manX_manY_UTR","mdtA_5_UTR","mdtE_5_UTR","mdtE_5_UTR","miaA_hfq_UTR","miaA_hfq_UTR","moaA_moaB_UTR","mreB_5_UTR","mscS_5_UTR","nhaA_nhaR_UTR","nnr_5_UTR","nnr_5_UTR","nsrR_rnr_UTR","nsrR_rnr_UTR","nuoC_nuoB_UTR","nuoC_nuoB_UTR","ompR_5_UTR","ompR_5_UTR","oppA_5_UTR","oppA_oppB_UTR","pck_5_UTR","pgaA_5_UTR","pgaA_5_UTR","phoU_5_UTR","phoU_5_UTR","phoU_5_UTR","pqqL_yddB_UTR","proB_5_UTR","proP_5_UTR","pspA_5_UTR","purE_5_UTR","purE_5_UTR","purL_5_UTR","purL_5_UTR","ribB_5_UTR","ribB_5_UTR","rnk_5_UTR","rnk_5_UTR","rpoS_5_UTR","rsmE_gshB_UTR","rsmE_gshB_UTR","sdhD_sdhA_UTR","sdiA_5_UTR","seqA_pgm_UTR","skp_5_UTR","sucA_5_UTR","sucC_5_UTR","talA_5_UTR","tauC_tauD_UTR","tnaA_5_UTR","topA_5_UTR","truC_5_UTR","uidR_5_UTR","uspG_5_UTR","uxaB_5_UTR","uxaC_5_UTR","uxaC_5_UTR","uxuA_5_UTR","yadM_htrE_UTR","yafK_yafQ_UTR","yafQ_5_UTR","yajI_5_UTR","ybaL_5_UTR","ybaL_5_UTR","ybaL_5_UTR","ybjP_5_UTR","ycaC_5_UTR","ycaK_5_UTR","ycdT_5_UTR","ycdT_5_UTR","ycdT_5_UTR","ydcJ_5_UTR","ydeP_5_UTR","ydhQ_5_UTR","ydhQ_5_UTR","yeaH_yeaI_UTR","yeaY_5_UTR","yeaY_5_UTR","yecC_5_UTR","ygjV_uxaA_UTR","yhbO_5_UTR","yhbO_5_UTR","yhcB_5_UTR","yhhQ_dcrB_UTR","yhhQ_dcrB_UTR","yiaD_5_UTR","yiaD_5_UTR","yiaD_ghrB_UTR","yidQ_5_UTR","yidQ_5_UTR","yiiS_5_UTR","yjbD_5_UTR"]

# ── Colors (light/white theme) ─────────────────────────────────────────────────
PROBE_COLORS = {
    'up':       (0.11, 0.62, 0.38, 0.95),   # green  — significant, FC >= 1
    'down':     (0.85, 0.18, 0.15, 0.95),   # red    — significant, FC <  1
    'ns_reads': (0.65, 0.65, 0.65, 0.90),   # medium gray — not significant, enough reads
    'no_reads': (0.30, 0.30, 0.38, 0.90),   # dark slate — not enough reads
}

# Figure / axes backgrounds
FIG_FACECOLOR  = '#ffffff'
AX_FACECOLOR   = '#f5f5f8'
SPINE_COLOR    = '#bbbbcc'
TICK_COLOR     = '#444455'
TEXT_COLOR     = '#111122'
LABEL_COLOR    = '#555566'

MEAN_LINE_COLOR = '#c49a00'   # darker gold — readable on white
MEAN_LINE_WIDTH = 1.6
BASELINE_COLOR  = '#888899'
SEGMENT_LW      = 2.2
LABEL_FONTSIZE  = 6.0
GGA_SIZE        = 20
GGA_MARKER = '.'
GGA_COLOR = 'blue'
GGA_EDGE_W = 1

# Extra right margin (in axes-fraction units) reserved for the μ annotation
RIGHT_MARGIN_FRAC = 0.10


def probe_color(row):
    if row['significant']:
        return PROBE_COLORS['up'] if row['fold_change'] >= 1 else PROBE_COLORS['down']
    elif row['enough_reads']:
        return PROBE_COLORS['ns_reads']
    else:
        return PROBE_COLORS['no_reads']


def assign_lanes(target_df):
    """Greedy lane packing so overlapping probes don't collide visually."""
    sorted_df = target_df.sort_values('start').copy()
    lanes, lane_ids = [], []
    for _, row in sorted_df.iterrows():
        placed = False
        for i, lane_end in enumerate(lanes):
            if row['start'] >= lane_end:
                lanes[i] = row['end']
                lane_ids.append(i)
                placed = True
                break
        if not placed:
            lanes.append(row['end'])
            lane_ids.append(len(lanes) - 1)
    sorted_df['lane'] = lane_ids
    return sorted_df


def build_figure(chunk, groups, start_codon_map):
    """Create and return a fully-rendered figure for a chunk of targets.

    `groups`          : dict {target -> probe DataFrame} (precomputed once).
    `start_codon_map` : dict {target -> start codon position} (precomputed once).
    """
    fig, axes = plt.subplots(
        len(chunk), 1,
        figsize=(14, len(chunk) * 3.4 + 1),
        squeeze=False,
    )
    fig.patch.set_facecolor(FIG_FACECOLOR)

    for ax, target in zip(axes[:, 0], chunk):
        target_df = assign_lanes(groups[target])

        vals = target_df['fold_change'].dropna()
        global_ymax = (vals.max() if len(vals) else 2) * 1.12
        global_ymin = 0.0

        LANE_JITTER = (global_ymax - global_ymin) * 0.018

        x_min   = target_df['start'].min()
        x_max   = target_df['end'].max()
        x_span  = x_max - x_min or 1
        padding = x_span * 0.03 or 20

        x_right_extra = x_span * RIGHT_MARGIN_FRAC

        ax.set_facecolor(AX_FACECOLOR)
        for spine in ax.spines.values():
            spine.set_color(SPINE_COLOR)

        ax.set_xlim(x_min - padding, x_max + padding + x_right_extra)
        ax.set_ylim(global_ymin, global_ymax)

        # ── baseline at FC = 1 ────────────────────────────────────────────────
        ax.axhline(1.0, color=BASELINE_COLOR, linewidth=0.9,
                   linestyle='--', zorder=1)
        ax.text(
            x_min - padding, 1.0,
            'FC=1', ha='right', va='center',
            fontsize=5.5, color=BASELINE_COLOR,
        )

        # ── start codon vertical line ─────────────────────────────────────────
        start_codon = start_codon_map.get(target)
        if start_codon is not None and not pd.isna(start_codon):
            ax.axvline(start_codon, linewidth=1.5,
                       color='#cc2222', zorder=4, alpha=0.9)

        # ── significance mean line ────────────────────────────────────────────
        sig_df   = target_df[target_df['significant'] == True]
        sig_vals = sig_df['fold_change'].dropna()
        if len(sig_vals):
            mean_fc = sig_vals.mean()
            ax.hlines(
                mean_fc,
                xmin=x_min - padding,
                xmax=x_max + padding,
                colors=MEAN_LINE_COLOR,
                linewidths=MEAN_LINE_WIDTH,
                linestyles='-.',
                zorder=3,
                alpha=0.90,
            )
            ax.annotate(
                f'μ={mean_fc:.3f}',
                xy=(x_max + padding, mean_fc),
                xycoords='data',
                xytext=(4, 0),
                textcoords='offset points',
                va='center', ha='left',
                fontsize=7, color=MEAN_LINE_COLOR,
                fontweight='bold', zorder=6,
                annotation_clip=False,
            )

        # ── draw probes as horizontal segments ────────────────────────────────
        for _, row in target_df.iterrows():
            fc = row['fold_change']
            if pd.isna(fc):
                continue

            color     = probe_color(row)
            probe_num = row['probename'].split('_probe_')[-1]
            lane      = row['lane']
            y_pos     = fc + lane * LANE_JITTER

            ax.hlines(
                y_pos,
                xmin=row['start'],
                xmax=row['end'],
                colors=color,
                linewidths=SEGMENT_LW,
                zorder=2,
            )
            ax.scatter(
                [row['start'], row['end']],
                [y_pos, y_pos],
                s=18, color=color, zorder=3, linewidths=0,
            )
            bonus_label = ''
            if row['Hfq_CLASH_Overlap']:
                bonus_label = bonus_label + 'H'
            if row['cut_site']:
                bonus_label = bonus_label + ' R '
            if row['realBS']:
                bonus_label = bonus_label + ' FPBS '
            midpoint = (row['start'] + row['end']) / 2
            fulllabel = probe_num
            if bonus_label:
                fulllabel = fulllabel + '(' + bonus_label + ')'
            ax.text(
                midpoint,
                y_pos + global_ymax * 0.012,
                fulllabel,
                ha='center', va='bottom',
                fontsize=LABEL_FONTSIZE,
                color=TEXT_COLOR,
                zorder=4,
            )

            if row['either_OR']:
                ax.scatter(
                    midpoint, y_pos,
                    s=GGA_SIZE, marker=GGA_MARKER,
                    color=GGA_COLOR,
                    edgecolors='white', linewidths=GGA_EDGE_W,
                    zorder=5,
                )

        # ── axes cosmetics ────────────────────────────────────────────────────
        ax.tick_params(colors=TICK_COLOR, labelsize=8)
        for lbl in ax.get_xticklabels():
            lbl.set_color(TICK_COLOR)
        for lbl in ax.get_yticklabels():
            lbl.set_color(TICK_COLOR)

        # Target name on left
        ax.text(
            -0.005, 0.5, target,
            transform=ax.transAxes,
            rotation=0, ha='right', va='center',
            fontsize=9, color=TEXT_COLOR, fontweight='bold',
        )
        ax.yaxis.set_label_position('right')
        ax.set_ylabel('Fold Change', color=LABEL_COLOR, fontsize=8, labelpad=6)

    # ── legend ────────────────────────────────────────────────────────────────
    legend_handles = [
        mlines.Line2D([], [], color=PROBE_COLORS['up'],       linewidth=2,
                      label='Significant ↑ (FC ≥ 1)'),
        mlines.Line2D([], [], color=PROBE_COLORS['down'],     linewidth=2,
                      label='Significant ↓ (FC < 1)'),
        mlines.Line2D([], [], color=PROBE_COLORS['ns_reads'], linewidth=2,
                      label='Not Significant (enough reads)'),
        mlines.Line2D([], [], color=PROBE_COLORS['no_reads'], linewidth=2,
                      label='Not Enough Reads'),
        mlines.Line2D([], [], color=MEAN_LINE_COLOR, linewidth=MEAN_LINE_WIDTH,
                      linestyle='-.', label='Mean FC (significant probes)'),
        mlines.Line2D([], [], color='#cc2222', linewidth=1.5,
                      label='Start Codon'),
    ]
    fig.legend(
        handles=legend_handles,
        loc='upper right', framealpha=0.5,
        labelcolor=TEXT_COLOR, facecolor=FIG_FACECOLOR,
        edgecolor=SPINE_COLOR, fontsize=6.5, ncols=2,
    )

    fig.text(0.5, 0.005, 'Transcript Position', ha='center',
             color=LABEL_COLOR, fontsize=10)
    fig.suptitle('Probe Coverage by Target (Raw Fold Change)', color=TEXT_COLOR,
                 fontsize=14, fontweight='bold')
    plt.tight_layout(h_pad=0.8)
    return fig


def _write_targets_pdf(targets, groups, start_codon_map,
                       output_pdf, targets_per_page):
    """Render a list of targets into a multi-page PDF, chunked per page."""
    target_chunks = [targets[i:i + targets_per_page]
                     for i in range(0, len(targets), targets_per_page)]
    with PdfPages(output_pdf) as pdf:
        for chunk in target_chunks:
            fig = build_figure(chunk, groups, start_codon_map)
            pdf.savefig(fig, facecolor=fig.get_facecolor(), bbox_inches='tight')
            plt.close(fig)
    print(f"Saved → {output_pdf}  ({len(targets)} targets)")


def summarize_fold_change(df, flagged_set):
    """Compute and print average fold change for flagged vs non-flagged targets.

    Returns a DataFrame indexed by group ('flagged' / 'non-flagged') with:
      n_targets        — distinct targets in the group
      n_probes         — probes with a valid fold change
      mean_fc_all      — mean FC pooled over every probe (probe-weighted)
      mean_fc_by_target— mean of each target's mean FC (target-weighted)
      n_sig_probes     — significant probes (pval < 0.05)
      mean_fc_sig      — mean FC pooled over significant probes only
    """
    valid = df[df['target'].notna() & df['fold_change'].notna()].copy()
    valid['group'] = np.where(
        valid['target'].isin(flagged_set), 'flagged', 'non-flagged'
    )

    has_sig = 'significant' in valid.columns

    rows = []
    for group_name, gdf in valid.groupby('group'):
        sig = gdf[gdf['significant'] == True] if has_sig else gdf.iloc[0:0]
        rows.append({
            'group':             group_name,
            'n_targets':         gdf['target'].nunique(),
            'n_probes':          len(gdf),
            'mean_fc_all':       gdf['fold_change'].mean(),
            'mean_fc_by_target': gdf.groupby('target')['fold_change'].mean().mean(),
            'n_sig_probes':      len(sig),
            'mean_fc_sig':       sig['fold_change'].mean() if len(sig) else float('nan'),
        })

    summary = pd.DataFrame(rows).set_index('group')

    print("\n── Average fold change summary ─────────────────────────────────")
    print(summary.to_string(float_format=lambda x: f'{x:.4f}'))
    print("────────────────────────────────────────────────────────────────\n")
    return summary


def visualize_probes(df, utr_data,
                     output_pdf='probe_visualization_vFinal_11_3.pdf',
                     flagged_output_pdf='figure_s5.pdf',
                     targets_per_page=6,
                     flagged_per_page=1,
                     flagged_order='list'):
    """
    Writes two PDFs:
      • flagged_output_pdf — only the targets in FLAGGED_TARGETS
      • output_pdf         — everything else (flagged targets excluded)

    Parameters
    ----------
    targets_per_page : targets per page in the main PDF.
    flagged_per_page : targets per page in the flagged PDF (1 = one per page).
    flagged_order    : 'list' keeps the order typed in FLAGGED_TARGETS;
                       'data' keeps the order targets appear in the data.
    """
    df = df.copy()
    df['target'] = df['probename'].str.extract(r'^(.+?)_probe_\d+$')
    df['fold_change'] = pd.to_numeric(df['fold_change'], errors='coerce')

    # Precompute per-target probe groups ONCE instead of re-scanning df per
    # target with a boolean mask. Turns O(n_targets * n_rows) into O(n_rows).
    groups = {t: g for t, g in df.groupby('target')}

    # Precompute target → start codon lookup ONCE (first occurrence wins),
    # replacing a full utr_data scan per target.
    start_codon_map = (
        utr_data.drop_duplicates('target')
                .set_index('target')['start_codon']
                .to_dict()
    )

    # Drop any probenames that didn't match the regex (target is NaN).
    targets     = [t for t in df['target'].unique() if pd.notna(t)]
    target_set  = set(targets)
    flagged_set = set(FLAGGED_TARGETS)

    # Flagged targets, deduped, only those actually present in the data.
    if flagged_order == 'data':
        flagged_targets = [t for t in targets if t in flagged_set]
    else:  # 'list' — order they were typed in FLAGGED_TARGETS
        seen = set()
        flagged_targets = []
        for t in FLAGGED_TARGETS:
            if t in target_set and t not in seen:
                flagged_targets.append(t)
                seen.add(t)

    # Everything else, in original data order.
    main_targets = [t for t in targets if t not in flagged_set]

    # ── flagged-only PDF ──────────────────────────────────────────────────────
    if flagged_targets:
        _write_targets_pdf(flagged_targets, groups, start_codon_map,
                           flagged_output_pdf, flagged_per_page)
        missing = flagged_set - target_set
        if missing:
            print(f"  Note: flagged targets not found in data: {sorted(missing)}")
    else:
        print("No flagged targets found in data — skipping flagged PDF.")

    # ── main PDF (flagged targets excluded) ───────────────────────────────────
    _write_targets_pdf(main_targets, groups, start_codon_map,
                       output_pdf, targets_per_page)

    # ── average fold change summary (flagged vs non-flagged) ──────────────────
    return summarize_fold_change(df, flagged_set)


# ── Entry point ───────────────────────────────────────────────────────────────
if __name__ == "__main__":
    UTR_data = pd.read_excel('../all_5_utrs_for_start_codon.xlsx')
    UTR_data['target'] = UTR_data['UTR'].str.extract(r'^(.+?)_sequfile$')[0]
    utrs = UTR_data
    every_probe = pd.read_excel('everyprobe_w_needed_data-7-6.xlsx')
    every_probe['significant'] = every_probe['padj'] < 0.05
    UTR_data['start_codon'] = np.where(
        utrs['Direction'] == 'F',
        pd.to_numeric(utrs['Genomic Coords start codon location'], errors='coerce')
            - utrs['Left Genomic Coord'],
        pd.to_numeric(
            utrs['Right Genomic Coord'] - utrs['Genomic Coords start codon location'],
            errors='coerce'),
    )

    summary = visualize_probes(every_probe, UTR_data)
    # `summary` is a DataFrame you can save if you want it on disk, e.g.:
    # summary.to_csv('fold_change_summary.csv')
```
